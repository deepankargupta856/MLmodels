```python
from sklearn.datasets import load_diabetes
from sklearn.metrics import r2_score
import numpy as np
import pandas as pd
from sklearn.linear_model import SGDRegressor
from sklearn.model_selection import train_test_split
from sklearn.linear_model import Ridge
```

```python
X,y = load_diabetes(return_X_y=True)
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=4)
```

```python
reg = SGDRegressor(penalty='l2',max_iter=500,eta0=0.1,learning_rate='constant',alpha=0.001)
```

```python
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
print(reg.coef_)
print(reg.intercept_)
print('R2 score:' ,r2_score(y_test,y_pred))
```

```python
rreg = Ridge(alpha=0.001, max_iter=500,solver='sparse_cg') # iterative
```

```python
rreg.fit(X_train,y_train)

y_pred = rreg.predict(X_test)
print("R2 score",r2_score(y_test,y_pred))
print(rreg.coef_)
print(rreg.intercept_)
```

```python
class MyRidgeGD:
  def __init__(self,epochs,learning_rate,alpha=0.1):
    self.alpha = alpha
    self.epochs = epochs
    self.learning_rate = learning_rate
    self.coef = None
    self.intercept = None

  def fit(self,X_train,Y_train):
    self.coef = np.ones(X_train.shape[1])
    self.intercept = 0
    weights = np.insert(self.coef,0,self.intercept)
    X_train = np.insert(X_train,0,1,axis=1)
    for i in range(self.epochs):
      derivative =  np.dot(X_train.T,X_train).dot(weights) - np.dot(X_train.T,y_train) + self.alpha*weights
      weights = weights - self.learning_rate*derivative

    self.intercept = weights[0]
    self.coef = weights[1:]

  def predict(self,X_test):
    return np.dot(X_test,self.coef) + self.intercept
```

```python
rgd = MyRidgeGD(epochs=500,alpha=0.001,learning_rate=0.005)
```

```python
rgd.fit(X_train,y_train)

y_pred = rgd.predict(X_test)
print("R2 score",r2_score(y_test,y_pred))
print(rgd.coef)
print(rgd.intercept)
```
